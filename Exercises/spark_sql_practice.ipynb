```python
%sql
DROP TABLE IF EXISTS flights_csv;
CREATE TABLE flights_csv AS
SELECT 
    DEST_COUNTRY_NAME, 
    ORIGIN_COUNTRY_NAME,
    count
FROM read_files(
    '/Volumes/workspace/default/flight_data/2015-summary.csv',
    format => 'csv',
    header => true
)
```

```python
%sql
CREATE TABLE IF NOT EXISTS flights_from_select AS SELECT * FROM flights_csv
```

```python
%sql
DROP TABLE IF EXISTS flights_csv_partitioned;
CREATE TABLE flights_csv_partitioned
PARTITIONED BY (DEST_COUNTRY_NAME)
AS
SELECT 
    DEST_COUNTRY_NAME, 
    ORIGIN_COUNTRY_NAME, 
    count
FROM read_files(
    '/Volumes/workspace/default/flight_data/2015-summary.csv',
    format => 'csv',
    header => true
);

SHOW PARTITIONS flights_csv_partitioned;
```
